## Local Ollama fine-tune workflow

This notebook has been converted to a local Ollama workflow.

Important: Ollama is best for local model serving and custom model creation from a base model. It does not support the same Hugging Face DPO/LoRA training pipeline used in the original notebook. For a local setup, use Ollama with a `Modelfile` and a base model such as `llama3.2:3b`.


````python
import json
import subprocess
from pathlib import Path

print("Local Ollama fine-tuning workflow ready.")
````

````python
base_model = "llama3.2:3b"  # use a local Ollama base model
````

````python
from pathlib import Path

# Create a small local training dataset in Ollama-friendly JSONL format
training_data = [
    {"input": "What is insulin used for?", "output": "Insulin is a hormone used to help the body regulate blood sugar and is commonly used in diabetes treatment."},
    {"input": "Explain how Metformin works.", "output": "Metformin lowers glucose production in the liver and improves insulin sensitivity, which can help manage type 2 diabetes."},
    {"input": "Why are clinical trials important?", "output": "Clinical trials test safety and effectiveness, helping researchers decide whether a treatment is suitable for patient care."},
]

train_path = Path("./local_ollama_training_data.jsonl")
with train_path.open("w", encoding="utf-8") as f:
    for row in training_data:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print(f"Created dataset at {train_path.resolve()}")
````

````python
from pathlib import Path

modelfile = Path("./Modelfile")
modelfile.write_text(
    '''FROM llama3.2:3b
SYSTEM """
You are a helpful pharmaceutical AI assistant.
Answer clearly, clinically cautious, and in a concise but informative way.
"""

TEMPLATE """
{{ if .System }}{{ .System }}\n{{ end }}
{{ .Prompt }}
"""
''',
    encoding="utf-8",
)

print(f"Modelfile created at {modelfile.resolve()}")
````

````python
local_model_name = "pharma-local-model"
````

````python
# Build the local Ollama model from the Modelfile.
# Run this in your terminal if Ollama is installed locally:
# ollama create pharma-local-model -f ./Modelfile

print("Run: ollama create pharma-local-model -f ./Modelfile")
````

````python
prompt = "Explain how Metformin works in the human body and why some researchers believe it could have benefits beyond diabetes treatment."
````

````python
# Use Ollama locally instead of CUDA Hugging Face generation.
print("Local Ollama inference command: ollama run pharma-local-model \"" + prompt + "\"")
````

````python
# This notebook now uses a local Ollama model instead of Hugging Face `generate()`.
# The old `instruction_model` variable no longer exists in this local workflow.
print(f"Run locally with: ollama run {local_model_name} \"{prompt}\"")
````

````python
print("To test the local model, run the following in a terminal:")
print("ollama run pharma-local-model \"Explain how Metformin works in the human body and why some researchers believe it could have benefits beyond diabetes treatment.\"")
````

```bash

ollama run pharma-local-model
```

For Ollama, the pattern is:

- `FROM ...` to choose the base model
- `SYSTEM ...` to define behavior
- `PARAMETER temperature ...` to tune generation
- `ollama create ...` to build a local model
- `ollama run ...` to test it

This replaces the Hugging Face `get_peft_model()` / `PeftModel.from_pretrained()` flow.
